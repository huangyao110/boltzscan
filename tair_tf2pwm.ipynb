```python
# from cropd2p.cropdock import crop_seq
# from Bio import SeqIO
# from utils.io_utils import *
import pandas as pd
from pathlib import Path
import json
```

```python
meme_dir = Path("data/pwms/tair/pwms_all_memes")
tair_tf2pwms = pd.read_csv("data/pwms/tair/TF_Information_all_motifs_plus.txt", sep="\t")
```

```python
tair_tf2pwms_dct = dict(zip(tair_tf2pwms["DBID"], tair_tf2pwms["Motif_ID"]))
from collections import defaultdict

tf2pwms_dct = defaultdict(list)

for dbid, motif in zip(tair_tf2pwms["DBID"], tair_tf2pwms["Motif_ID"]):
    tf2pwms_dct[dbid].append(motif)
```

```python
with open("data/pwms/tair/tf2pwms_dct.json", "w") as f:
    json.dump(tf2pwms_dct, f, indent=4)
```

```python
def reverse_kv(dct):
    new_dct = {}
    for key, value in dct.items():
        # 如果 value 不可哈希，这里会自然抛异常，行为明确
        new_dct.setdefault(value, []).append(key)
    return new_dct
pwms2tf_dct = reverse_kv(tair_tf2pwms_dct)
```
